```python
import torch
```

```python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(

            # 1st hidden layer
            torch.nn.Linear(num_inputs, 100),
            torch.nn.ReLU(),

            # 2nd hidden layer
            torch.nn.Linear(100, 250),
            torch.nn.ReLU(),

            # output layer
            torch.nn.Linear(250, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits


model = NeuralNetwork(100, 5)
print(model)
```
