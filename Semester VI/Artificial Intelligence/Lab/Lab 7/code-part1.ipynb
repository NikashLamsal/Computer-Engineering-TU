### A.2.1 Scalars, vectors, matrices, and tensors

````python
import torch
import numpy as np

# create a 0D tensor (scalar) from a Python integer
tensor0d = torch.tensor(1)

# create a 1D tensor (vector) from a Python list
tensor1d = torch.tensor([1, 2, 3])

# create a 2D tensor from a nested Python list
tensor2d = torch.tensor([[1, 2], 
                         [3, 4]])

# create a 3D tensor from a nested Python list
tensor3d_1 = torch.tensor([[[1, 2], [3, 4]], 
                           [[5, 6], [7, 8]]])

# create a 3D tensor from NumPy array
ary3d = np.array([[[1, 2], [3, 4]], 
                  [[5, 6], [7, 8]]])
tensor3d_2 = torch.tensor(ary3d)  # Copies NumPy array
tensor3d_3 = torch.from_numpy(ary3d)  # Shares memory with NumPy array
````

## A.3 Seeing models as computation graphs

##  Automatic Differentiation Engine also known as **autograd**

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/7.webp" width="600px">

````python
import torch.nn.functional as F

y = torch.tensor([1.0])  # true label
x1 = torch.tensor([1.1]) # input feature
w1 = torch.tensor([2.2]) # weight parameter
b = torch.tensor([0.0])  # bias unit

z = x1 * w1 + b          # net input
a = torch.sigmoid(z)     # activation & output

loss = F.binary_cross_entropy(a, y)
print(loss)
````

## A.4 Automatic differentiation made easy

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/8.webp" width="600px">

## By default, PyTorch destroys the computation graph after calculating the gradients to free memory. 
## However, since we will reuse this computation graph shortly,we set **retain_graph=True** so that it stays in memory

````python
import torch.nn.functional as F
from torch.autograd import grad

y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.2], requires_grad=True)
b = torch.tensor([0.0], requires_grad=True)

z = x1 * w1 + b 
a = torch.sigmoid(z)

loss = F.binary_cross_entropy(a, y)
print(loss)
````

````python
grad_L_w1 = grad(loss, w1, retain_graph=True)
grad_L_b = grad(loss, b, retain_graph=True)

print(grad_L_w1)
print(grad_L_b)
````

````python
loss.backward()

print(w1.grad)
print(b.grad)
````

## Code disections
```bash
1. class NeuralNetwork(torch.nn.Module):
### Details
Automatically gains the ability to track weights, biases, move to GPUs, and calculate gradients for training.
```
- - -

```bash
2. def __init__(self, num_inputs, num_outputs):
    super().__init__()

### Details
__init__: This initializes the network and defines its architecture (the "blueprint"). It accepts the size of the incoming data (num_inputs) and the size of the desired prediction (num_outputs).

super().__init__(): This is a crucial line in Python. It triggers the initialization of the parent class (torch.nn.Module). Without this, PyTorch won't register the layers you define below it.
```
- - -

```bash
self.layers = torch.nn.Sequential()

torch.nn.Sequential: A convenient wrapper. Instead of passing data manually from layer to layer, Sequential chains them together so that the output of one automatically becomes the input to the next.

```
- - -
```bash
torch.nn.Linear(num_inputs, 30),
torch.nn.ReLU(),


### Details

1. Linear(num_inputs, 30): This is a fully connected layer. It takes num_inputs features (e.g., 10 features of a house) and applies a linear transformation ($Y = XW^T + b$) to project them into a 30-dimensional space.

2. ReLU(): The Rectified Linear Unit activation function. It replaces all negative numbers coming out of the linear layer with 0. Why? It introduces non-linearity, allowing the network to learn complex, curvy patterns instead of just straight lines.

```
- - -

```bash
torch.nn.Linear(20, num_outputs),


The final stop. It maps the 20 hidden features down to the exact number of outputs you need (e.g., 3 outputs if you are classifying images into cats, dogs, or birds). Notice there is no ReLU here, as we want raw, unconstrained values.
```

- - -

```bash
def forward(self, x):
    logits = self.layers(x)
    return logits


forward(self, x): This defines the actual flow of data. When you pass data x into the network, Python executes this method.

logits: The raw, unnormalized outputs generated by the final linear layer.

return logits: Sends the predictions out so you can calculate the error (loss) and update the network.
```
- - -

````python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super().__init__()

        self.layers = torch.nn.Sequential(
                
            # 1st hidden layer
            torch.nn.Linear(num_inputs, 30),
            torch.nn.ReLU(),

            # 2nd hidden layer
            torch.nn.Linear(30, 20),
            torch.nn.ReLU(),

            # output layer
            torch.nn.Linear(20, num_outputs),
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits
````

````python
model = NeuralNetwork(50, 3)
````

````python
print(model)
````

a Uniform Distribution ($\mathcal{U}$) bounded by a specific range:

$$\text{Range} = \left[ -\sqrt{k}, +\sqrt{k} \right]$$

Where:
$$k = \frac{1}{\text{in\_features}}$$

Both the weights and the biases are initialized using this exact same range by default.

## A.6 Setting up efficient data loaders

<img src="https://sebastianraschka.com/images/LLMs-from-scratch-images/appendix-a_compressed/10.webp" width="600px">

````python
X_train = torch.tensor([
    [-1.2, 3.1],
    [-0.9, 2.9],
    [-0.5, 2.6],
    [2.3, -1.1],
    [2.7, -1.5]
])

y_train = torch.tensor([0, 0, 0, 1, 1])
````

````python
X_test = torch.tensor([
    [-0.8, 2.8],
    [2.6, -1.6],
])

y_test = torch.tensor([0, 1])
````

````python
from torch.utils.data import Dataset


class ToyDataset(Dataset):
    def __init__(self, X, y):
        self.features = X
        self.labels = y

    def __getitem__(self, index):
        one_x = self.features[index]
        one_y = self.labels[index]        
        return one_x, one_y

    def __len__(self):
        return self.labels.shape[0]

train_ds = ToyDataset(X_train, y_train)
test_ds = ToyDataset(X_test, y_test)
````

````python
train_ds[0]
````

````python
len(train_ds)
````

````python
from torch.utils.data import DataLoader
# DataLoader: parallelizing data loading and preprocessing
# num_workers: 0 the data loading will be done in the main process and not in separate worker processes.
torch.manual_seed(123)

train_loader = DataLoader(
    dataset=train_ds, # 
    batch_size=2,
    shuffle=True,
    num_workers=0
)
````

````python
test_ds = ToyDataset(X_test, y_test)

test_loader = DataLoader(
    dataset=test_ds,
    batch_size=2,
    shuffle=False,
    num_workers=0
)
````

````python
for idx, (x, y) in enumerate(train_loader):
    print(f"Batch {idx+1}:", x, y)
````

````python
train_loader = DataLoader(
    dataset=train_ds,
    batch_size=2,
    shuffle=True,
    num_workers=0,
    drop_last=True
)
````

## A.7 A typical training loop

````python
import torch.nn.functional as F


torch.manual_seed(123)
model = NeuralNetwork(num_inputs=2, num_outputs=2)
optimizer = torch.optim.SGD(model.parameters(), lr=0.5)

num_epochs = 3

for epoch in range(num_epochs):
    
    model.train()
    for batch_idx, (features, labels) in enumerate(train_loader):

        logits = model(features)
        
        loss = F.cross_entropy(logits, labels) # Loss function
        
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    
        ### LOGGING
        print(f"Epoch: {epoch+1:03d}/{num_epochs:03d}"
              f" | Batch {batch_idx+1:03d}/{len(train_loader):03d}"
              f" | Train/Val Loss: {loss:.2f}")

    model.eval()
    # Optional model evaluation
````

````python
def compute_accuracy(model, dataloader):

    model.eval()
    correct = 0.0
    total_examples = 0
    
    for idx, (features, labels) in enumerate(dataloader):
        
        with torch.no_grad():
            logits = model(features)
        
        predictions = torch.argmax(logits, dim=1)
        compare = labels == predictions
        correct += torch.sum(compare)
        total_examples += len(compare)

    return (correct / total_examples).item()
````

````python
compute_accuracy(model, train_loader)
````

````python
compute_accuracy(model, test_loader)
````
