## Theory

**Divide and Conquer** is an algorithm design paradigm that solves a problem by breaking it down into smaller subproblems of the same type, solving each subproblem recursively, and then combining the results to solve the original problem. It generally consists of three steps:

1. **Divide**: Break the given problem into smaller sub-problems.
2. **Conquer**: Solve the sub-problems recursively. If the sub-problem is small enough, solve it directly (base case).
3. **Combine**: Combine the solutions of the sub-problems to get the solution of the original problem.

Common examples of divide-and-conquer algorithms include Binary Search, Merge Sort, Quick Sort, and the divide-and-conquer approach to finding the minimum and maximum of an array.

**Heap Sort**, although not strictly a divide-and-conquer algorithm, is included here as a comparison sorting algorithm. It uses a binary heap data structure to repeatedly extract the maximum (or minimum) element and build a sorted array, running in O(n log n) time.

**Time Complexity Overview**:

| Algorithm | Best Case | Average Case | Worst Case |
|---|---|---|---|
| Binary Search | O(1) | O(log n) | O(log n) |
| Min-Max (Iterative) | O(n) | O(n) | O(n) |
| Min-Max (Divide and Conquer) | O(n) | O(n) | O(n) |
| Merge Sort | O(n log n) | O(n log n) | O(n log n) |
| Heap Sort | O(n log n) | O(n log n) | O(n log n) |
| Quick Sort | O(n log n) | O(n log n) | O(n²) |

## Related Diagram

**Divide and Conquer recursion tree (example: Merge Sort on an 8-element array)**

```
                    [8 elements]
                   /            \
            [4 elements]      [4 elements]
             /      \          /      \
        [2 elem]  [2 elem]  [2 elem]  [2 elem]
         /  \       /  \      /  \      /  \
       [1] [1]    [1] [1]  [1] [1]   [1] [1]
```

The array is divided in half at each level (**Divide**), single elements are trivially "sorted" (**Conquer** — base case), and pairs are merged back together in sorted order as the recursion unwinds (**Combine**).

## Computer Code

### 1. Binary Search

````python
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

arr = [2, 5, 8, 12, 16, 23, 38, 45, 56, 72]
target = 23
result = binary_search(arr, target)
print(f"Array: {arr}")
print(f"Target: {target}")
print(f"Index found: {result}")
````

### 2. Min and Max of an Array - Iterative Process

````python
def min_max_iterative(arr):
    mn = mx = arr[0]
    comparisons = 0
    for x in arr[1:]:
        if x < mn:
            mn = x
        comparisons += 1
        if x > mx:
            mx = x
        comparisons += 1
    return mn, mx, comparisons

arr = [7, 2, 9, 4, 1, 8, 5, 3]
mn, mx, comps = min_max_iterative(arr)
print(f"Array: {arr}")
print(f"Minimum: {mn}, Maximum: {mx}")
print(f"Number of comparisons: {comps}")
````

### 3. Min and Max of an Array - Divide and Conquer

````python
def min_max_dc(arr, low, high):
    if low == high:
        return arr[low], arr[low]
    if high - low == 1:
        if arr[low] < arr[high]:
            return arr[low], arr[high]
        else:
            return arr[high], arr[low]
    mid = (low + high) // 2
    min1, max1 = min_max_dc(arr, low, mid)
    min2, max2 = min_max_dc(arr, mid + 1, high)
    return min(min1, min2), max(max1, max2)

arr = [7, 2, 9, 4, 1, 8, 5, 3]
mn, mx = min_max_dc(arr, 0, len(arr) - 1)
print(f"Array: {arr}")
print(f"Minimum: {mn}, Maximum: {mx}")
````

### 4. Merge Sort

````python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

arr = [5, 2, 9, 1, 5, 6, 7, 3]
print(f"Original array: {arr}")
sorted_arr = merge_sort(arr)
print(f"Sorted array: {sorted_arr}")
````

### 5. Heap Sort

````python
def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heap_sort(arr):
    n = len(arr)
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)
    for i in range(n - 1, 0, -1):
        arr[0], arr[i] = arr[i], arr[0]
        heapify(arr, i, 0)
    return arr

arr = [12, 11, 13, 5, 6, 7]
print(f"Original array: {arr}")
sorted_arr = heap_sort(arr)
print(f"Sorted array: {sorted_arr}")
````

### 6. Quick Sort

````python
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def quick_sort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)
        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)

arr = [10, 7, 8, 9, 1, 5]
print(f"Original array: {arr}")
quick_sort(arr, 0, len(arr) - 1)
print(f"Sorted array: {arr}")
````

## Analysis of the Algorithms

- **Binary Search**: Runs in O(log n) time by repeatedly halving the search space, but requires the array to be sorted beforehand.
- **Min-Max (Iterative)**: Requires roughly **2n** comparisons since each element is compared once against the current minimum and once against the current maximum.
- **Min-Max (Divide and Conquer)**: Requires only about **3n/2** comparisons, since pairs of elements are first compared with each other, and only one of the two is then compared against the running min/max — reducing total comparisons.
- **Merge Sort**: Guarantees O(n log n) in all cases (best, average, worst), but requires O(n) extra space for merging.
- **Heap Sort**: Also O(n log n) in all cases and sorts **in-place** (O(1) extra space), unlike Merge Sort.
- **Quick Sort**: O(n log n) on average, but degrades to O(n²) in the worst case (e.g., when the array is already sorted and the last element is always chosen as pivot).
