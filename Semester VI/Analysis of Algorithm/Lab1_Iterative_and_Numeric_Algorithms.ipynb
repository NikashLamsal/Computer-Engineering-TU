## Theory

**Iterative Algorithms** solve a problem by repeating a set of steps (loops) until a condition is met, without necessarily breaking the problem into recursive sub-problems.

- **Sequential Search** scans a list one element at a time from the beginning until the target is found or the list ends. It works on both sorted and unsorted data, with O(n) time complexity.
- **Bubble Sort** repeatedly steps through the list, compares adjacent elements, and swaps them if they are in the wrong order. The largest unsorted element "bubbles up" to its correct position each pass. Time complexity: O(n²).
- **Insertion Sort** builds the sorted array one element at a time by taking each element and inserting it into its correct position among the previously sorted elements. Time complexity: O(n²), but efficient for nearly-sorted data.
- **Selection Sort** repeatedly selects the minimum element from the unsorted portion of the list and places it at the beginning. Time complexity: O(n²).

**Numeric Algorithms** deal with mathematical computations, often used in number theory and cryptography.

- **Euclid's Algorithm** computes the Greatest Common Divisor (GCD) of two numbers using the property `gcd(a, b) = gcd(b, a mod b)`, repeated until the remainder is zero.
- **Fibonacci Series** is a sequence where each number is the sum of the two preceding ones (0, 1, 1, 2, 3, 5, 8, ...), computable either recursively or iteratively.
- **Chinese Remainder Theorem (CRT)** provides a way to solve a system of simultaneous congruences with pairwise coprime moduli, uniquely determining a solution modulo the product of the moduli.
- **Fermat's Little Theorem** states that if `p` is a prime number, then for any integer `a` not divisible by `p`: `a^(p-1) ≡ 1 (mod p)`. This is widely used for primality testing and modular exponentiation.
- **Miller-Rabin Primality Test** is a probabilistic algorithm that determines whether a given number is likely prime by repeatedly testing a property that all prime numbers satisfy, using randomly chosen witnesses.

## Computer Code

### 1. GCD using Euclid's Algorithm

````python
def gcd_euclid(a, b):
    while b != 0:
        a, b = b, a % b
    return a

a, b = 48, 18
print(f"GCD of {a} and {b} is: {gcd_euclid(a, b)}")
````

### 2. Fibonacci Series

````python
def fibonacci_series(n):
    fib = [0, 1]
    for i in range(2, n):
        fib.append(fib[-1] + fib[-2])
    return fib[:n]

n = 10
print(f"First {n} terms of Fibonacci series:")
print(fibonacci_series(n))
````

### 3. Sequential Search

````python
def sequential_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

arr = [23, 11, 45, 8, 67, 34, 90]
target = 67
result = sequential_search(arr, target)
print(f"Array: {arr}")
print(f"Target: {target}")
print(f"Index found: {result}")
````

### 4. Bubble Sort

````python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

arr = [64, 34, 25, 12, 22, 11, 90]
print(f"Original array: {arr}")
sorted_arr = bubble_sort(arr)
print(f"Sorted array: {sorted_arr}")
````

### 5. Insertion Sort

````python
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

arr = [12, 11, 13, 5, 6]
print(f"Original array: {arr}")
sorted_arr = insertion_sort(arr)
print(f"Sorted array: {sorted_arr}")
````

### 6. Selection Sort

````python
def selection_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

arr = [64, 25, 12, 22, 11]
print(f"Original array: {arr}")
sorted_arr = selection_sort(arr)
print(f"Sorted array: {sorted_arr}")
````

### 7. Chinese Remainder Theorem

````python
def chinese_remainder_theorem(remainders, moduli):
    total = 0
    product = 1
    for m in moduli:
        product *= m

    for r_i, m_i in zip(remainders, moduli):
        p = product // m_i
        total += r_i * pow(p, -1, m_i) * p

    return total % product

# Example: x = 2 (mod 3), x = 3 (mod 5), x = 2 (mod 7)
remainders = [2, 3, 2]
moduli = [3, 5, 7]
result = chinese_remainder_theorem(remainders, moduli)
print(f"Remainders: {remainders}")
print(f"Moduli: {moduli}")
print(f"x = {result} (mod {3*5*7})")

# Verify
for r, m in zip(remainders, moduli):
    print(f"{result} mod {m} = {result % m} (expected {r})")
````

### 8. Fermat's (Little) Theorem

````python
def fermat_little_theorem_check(a, p):
    """
    Verifies Fermat's Little Theorem: a^(p-1) mod p == 1, if p is prime
    and a is not divisible by p.
    """
    if p < 2:
        return False
    result = pow(a, p - 1, p)
    return result == 1

def fermat_primality_test(n, k=5):
    """
    Probabilistic primality test based on Fermat's Little Theorem.
    """
    import random
    if n < 4:
        return n in (2, 3)
    for _ in range(k):
        a = random.randint(2, n - 2)
        if pow(a, n - 1, n) != 1:
            return False
    return True

a, p = 2, 7
print(f"Checking Fermat's Little Theorem for a={a}, p={p}")
print(f"{a}^{p-1} mod {p} = {pow(a, p - 1, p)}")
print(f"Holds (p is prime): {fermat_little_theorem_check(a, p)}")

print()
for n in [7, 15, 97, 100]:
    print(f"Is {n} prime (Fermat test)? {fermat_primality_test(n)}")
````

### 9. Miller-Rabin Primality Test

````python
import random

def miller_rabin(n, k=5):
    if n < 2:
        return False
    if n in (2, 3):
        return True
    if n % 2 == 0:
        return False

    # write n-1 as 2^r * d
    r, d = 0, n - 1
    while d % 2 == 0:
        r += 1
        d //= 2

    for _ in range(k):
        a = random.randint(2, n - 2)
        x = pow(a, d, n)
        if x == 1 or x == n - 1:
            continue
        for _ in range(r - 1):
            x = pow(x, 2, n)
            if x == n - 1:
                break
        else:
            return False
    return True

test_numbers = [7, 15, 97, 561, 7919, 1000000007]
for n in test_numbers:
    print(f"Is {n} prime (Miller-Rabin test)? {miller_rabin(n)}")
````

## Analysis of the Algorithms

- **Sequential Search**: O(n) time complexity; works on unsorted data but is inefficient for large datasets compared to Binary Search.
- **Bubble Sort**: O(n²) in the worst/average case, O(n) best case with the early-exit optimization (when the array is already sorted). Simple but inefficient for large inputs.
- **Insertion Sort**: O(n²) worst/average case, O(n) best case. Performs well on small or nearly-sorted datasets, and is used as the base case in hybrid sorts like Timsort.
- **Selection Sort**: O(n²) in all cases, since it always scans the remaining unsorted portion regardless of order. Makes the fewest swaps among the three (O(n) swaps), useful when write operations are costly.
- **GCD (Euclid's Algorithm)**: Runs in O(log(min(a, b))) time, extremely efficient compared to naive factorization-based methods.
- **Fibonacci Series**: The iterative version runs in O(n) time and O(1) extra space (excluding output storage), far more efficient than the naive recursive version, which is O(2^n).
- **Chinese Remainder Theorem**: Runs in O(k log M) time where k is the number of congruences and M is the product of the moduli; relies on modular inverses existing, which requires pairwise coprime moduli.
- **Fermat's Little Theorem**: The primality test runs in O(k log n) time for k rounds, but can be fooled by Carmichael numbers (e.g., 561), which pass the test despite being composite.
- **Miller-Rabin Test**: Also O(k log n) time, but is a stronger probabilistic test than Fermat's it correctly identifies Carmichael numbers like 561 as composite, making it the preferred primality test in practice.

## Discussion and Conclusion

This lab covered two categories of algorithms: simple iterative search/sort algorithms and classical numeric algorithms.

Among the sorting algorithms, all three (Bubble, Insertion, and Selection Sort) share O(n²) worst-case complexity, but they differ in practical behavior Insertion Sort adapts well to nearly-sorted input, Bubble Sort can terminate early when no swaps are needed, and Selection Sort minimizes the number of swaps at the cost of always performing the full comparison scan.

On the numeric side, Euclid's Algorithm demonstrated how a simple iterative approach can solve the GCD problem far more efficiently than brute-force methods. The Fibonacci series highlighted the importance of choosing an iterative approach over naive recursion to avoid exponential blow-up. The Chinese Remainder Theorem showed how modular arithmetic can be used to reconstruct a unique solution from a system of congruences, a technique widely used in cryptography (e.g., RSA optimizations).

Finally, comparing Fermat's Little Theorem with the Miller-Rabin test showed an important trade-off in primality testing: Fermat's test is simple but can be deceived by Carmichael numbers, while Miller-Rabin is more robust and is the standard choice for primality testing in real-world cryptographic applications.

Overall, this lab reinforced how both simple iterative techniques and deeper number-theoretic principles form the foundation of many practical algorithms used in computing today.
